# Quick Start

This notebook only confirms that the supplied files load correctly and use the
same anonymous station identifiers.

Read `README.md`, `rules_and_data_use.md`, and `data_dictionary.csv` before
starting analysis.


````python
from pathlib import Path
import pandas as pd

DATA_DIR = Path(".")

train = pd.read_parquet(DATA_DIR / "traffic_train.parquet")
validation = pd.read_parquet(DATA_DIR / "traffic_validation.parquet")
network = pd.read_csv(DATA_DIR / "network.csv")
edges = pd.read_csv(DATA_DIR / "network_edges.csv")
targets = pd.read_csv(DATA_DIR / "submission_template.csv")

print("train:", train.shape)
print("validation:", validation.shape)
print("network:", network.shape)
print("network edges:", edges.shape)
print("forecast targets:", targets.shape)
````

````python
print("Train dates:", train["timestamp"].min(), "to", train["timestamp"].max())
print("Validation dates:", validation["timestamp"].min(), "to", validation["timestamp"].max())

print("Train stations:", train["station_key"].nunique())
print("Validation stations:", validation["station_key"].nunique())
print("Network stations:", network["station_key"].nunique())
print("Target stations:", targets["station_key"].nunique())

joined = train.merge(network, on="station_key", how="left", validate="many_to_one")
print("Joined rows:", len(joined))
print("Station IDs missing from network:", joined["station_key"].isna().sum())
````
